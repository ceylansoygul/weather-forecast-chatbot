# Project Environment and Configuration

This notebook prepares the Python environment used by the project and loads the API configuration for the current notebook session.

For local use, create a `.env` file in the project root from `.env.example` and add your own API key. Keep `.env` private and out of version control. In Google Colab, provide the key through Colab Secrets.

```python
from pathlib import Path
from dotenv import load_dotenv
import os

project_root = Path.cwd().parent
env_file = project_root / ".env"

load_dotenv(env_file, override=True)

key = os.getenv("OPENAI_API_KEY")

print("ENV file exists:", env_file.exists())
print("API key found:", key is not None)
print("API key starts with sk-:", key.startswith("sk-") if key else False)
print("API key length:", len(key) if key else 0)
```

```python
import os
from pathlib import Path

print("Current folder:")
print(Path.cwd())

print("\n.env locations:")
for p in Path.cwd().parents:
    env_file = p / ".env"
    if env_file.exists():
        print("FOUND:", env_file)
```
